Os quatro tipos primitivos mais basicos que existem são:
*   int (Números inteiros)
*   float (NÚmeros reais ou numeros de ponto flutuante)
*   bool (Lógicos ou boleanos)
*   str (Caracteres ou strings)

Números inteiros (int)
ex: 7  -4  0  9875
todos esses são valores inteiros para o py.

Números reais (float)
ex: 4.5   0.076   -15.223  7.0

Valores lógicos ou booleanos
ex: True / False ou T / F (maiusculo)

Strings ou caracteres
ex: Olá  '7.5'  ''
